# Hospital Readmission Prediction Among Diabetic Patients

## 1.1 Research Problem
Hospital readmission is an important indicator of patient outcomes and healthcare quality. This project investigates
clinical and operational factors associated with hospital readmission among diabetic patients and develops predictive
models for identifying encounters associated with readmission within 30 days.

## 1.2 Research Questions
- Can clinical history, medication information, laboratory results and hospital encounter characteristics predict
30-day readmission?
- Which clinical and operational variables are most strongly associated with readmission?
- Does medication history improve prediction beyond demographic and admission-related information?
- How do different classification algorithms perform under class imbalance?
- Can the model generalise to later hospital encounters, and how do errors differ across demographic groups?
